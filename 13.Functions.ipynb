#### 8.Lambda Functions

* A lambda function in Python is a small, anonymous function defined using the lambda keyword.
* It can take any number of arguments, but is limited to a single expression, the result of which is automatically returned.
* SYNTAX: lambda arguments: expression
* lambda: The keyword used to define the anonymous function.
* arguments: One or more arguments, separated by commas (no parentheses are needed).
* expression: A single expression that is evaluated and returned. No return keyword is necessary. 


```python
# A lambda function that adds 10 to a given number
add_ten = lambda x: x + 10
print(add_ten(5)) # Output: 15
```

```python
# A lambda function with multiple arguments
multiply = lambda a, b: a * b
print(multiply(5, 6)) # Output: 30
```

```python
# Check if a number is even or odd
even_or_odd = lambda x: "Even" if x % 2 == 0 else "Odd"
print(even_or_odd(7)) # Output: Odd
```
