#### Input

→ input() -- get input from user 

--> input() always returns string (str).




```python
input("message")
```

```python
name = input("Enter your name: ")
print(name)
```

```python
print(type(name))  ### IMPORTANT  input() always returns string (str).
```

```python
# Converting Input to Integer
age = input("Age")
print(type(age))

age1 = int(input("Age1"))
print(type(age1))
```

```python
# Taking Multiple Inputs in One Line

a, b = input("Enter two numbers: ").split()

print(a, b)
print(a + b)

# Convert to integers:

a, b = map(int,input("Enter two numbers: ").split())

print(a, b)
print(a + b)
```

```python
name = input("Enter employee name: ")
salary = float(input("Enter salary: "))

print("Employee:", name)
print("Salary:", salary)
```
