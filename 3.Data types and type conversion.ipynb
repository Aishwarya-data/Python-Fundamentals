## MUTABLE OBJECTS:

* These objects can be modified after they are created
* We can change without creating new object
* Its memory location remains same EX - List,Dict,Set

```python
Lt = [1,2,3]

print(Lt)
print(id(Lt))

Lt.append(4)

print(Lt)
print(id(Lt))

#id memory address
```

# IMMUTABLE OBJECTS:

* cannot be modified once created
* Any operations that appears to modify the object actually creates a new object.
* Their original identity(memory location) remains constant but new value gets different identity.
*  EX - int,float , strings ,tuple

```python
str1 = "I am learning "
print(str1)
print(id(str1))

str1 += "Data science"
print(str1)
print(id(str1))
```

# Type Conversion

Type conversion --> one data type → another data type.

Two types:

* Implicit Conversion (Automatic)
* Explicit Conversion (Manual)

### 1.IMPLICIT CONVERSION:


```python
num1 = 5
num2 = 10.5

num3 = num1 + num2

print(num3)
print(type(num3))
```

### 2.Explicit Conversion

```python
#String → Integer

str1 = "123"

integer = int(str1)

print(integer)
print(type(integer))
```

```python
#Float → String

float1 = 3.9

str2 =str(float1)

print(str2)
print(type(str2))
```
