#### In Python, random numbers are generated using the built-in random module. It provides functions for generating random numbers, selecting random items, and shuffling data.

## Import module random

```python
import random
```

## Random integer  -- Generate a random integer between two numbers (inclusive):

```python
import random

num = random.randint(1, 10)
print(num)
```

## Random float (0–1)

```python
num = random.random()
print(num)
```

## Random number in a range

```python
num = random.uniform(1, 10)
print(num)
```

## Random choice from a list

```python
items = ["apple", "banana", "mango"]
print(random.choice(items))
```

## Shuffle a list

```python
numbers = [1, 2, 3, 4, 5]
random.shuffle(numbers)
print(numbers)
```

| Function         | Purpose                        |
| ---------------- | ------------------------------ |
| `random()`       | Random float between 0 and 1   |
| `randint(a,b)`   | Random integer between a and b |
| `uniform(a,b)`   | Random float between a and b   |
| `choice(list)`   | Pick random element            |
| `shuffle(list)`  | Shuffle list                   |
| `sample(list,n)` | Random unique items            |
