**What is OOP?**


Object-Oriented Programming is a way of writing code by grouping related data (attributes) and behaviour (methods) into reusable units called classes. Think of a class as a blueprint and an object as something built from that blueprint.

### 1. CLASS & OBJECTS:
* Class = Blueprint  |  Object = Instance
* A class defines the structure. You create objects (instances) from it. Each object has its own copy of the attributes.

* Object     =  The THING itself       → car1, ram
* Attribute  =  Details ABOUT the thing → .name, .brand, .color
* Method     =  Actions the thing does  → .drive(), .login()
* Inheritance — a child class gets everything from its parent. 


## REAL WORLD ANANLOGY:

I want to reuse the same function for Team Members and Managers, but they have small differences. How do I do that?

Both Team Members and Managers should be able to log in, but:

* Team Members → should log in, view tasks, and submit tasks.
* Managers → should log in, view tasks, and assign tasks.

*The Problem:*
* TeamMember  → login, view tasks, submit tasks
* Manager     → login, view tasks, assign tasks

* login and view tasks are the SAME for both.Only the last step is DIFFERENT.

### WITHOUT OOPS

```python
def team_member_work(name):
    print(f"{name} logged in")
    print(f"{name} viewing tasks")
    print(f"{name} submitted a task")  

def manager_work(name):
    print(f"{name} logged in")     
    print(f"{name} viewing tasks")  
    print(f"{name} assigned a task")
```

* Same code written twice!
* Imagine 10 roles — need to write 10 times

### WITH OOPS

#### Step 1 — Parent Class [Same part  -- login,view tasks]

```python
class user:
    def __init__(self,name):
        self.name=name
    def login(self):
        print(f"{self.name} logged in ")
    def view_tasks(self):
        print(f"{self.name} Viewed tasks")
        
```

#### Step 2 - Child class [ different part -- submit task for team memeber,assigned task for Manager]

```python
class Team_member(user):
    def work(self):
        self.login()
        self.view_tasks()
        print(f"{self.name} submitted tasks")
class Manager(user):
    def work(self):
        self.login()
        self.view_tasks()
        print(f"{self.name} Assigned tasks")
```

### step 3 OBJECTS CREATED  and CALLING METHODS

```python
#OBJECTS CREATED HERE!
user1 = Manager("Aishwarya_M")
user2 = Team_member("Lakshmi")
```

```python
##Calling the work() method on each object.
user1.work()
print("***************************************")
user2.work()
```

1. **class User:** creating a CLASS called User. A class is just a blueprint — nothing real yet.
2. **def __init__(self, name):** This is a special method called a constructor.
It runs automatically the moment you create an object.
name = the value you pass in like "Aishwarya"
3. **self.name = name** self = this specific object
self.name = save the name inside this object
So Aishwarya's name stays with Aishwarya

def login(self):
    print(f"{self.name} logged in ")
def view_tasks(self):
    print(f"{self.name} is viewing tasks ")

4. This is a **METHOD** — an action the object can do.
* method -- login,view_tasks  --- {self.name} picks up the name saved in this object.
* Both login and view_tasks belong to User class.
So every user can do these actions!


5. Creating a **new class** called **TeamMember**.
* (User) means — inherit everything from User!
* So TeamMember automatically gets login and view_tasks
* work() is TeamMember's own METHOD. It calls login and view_tasks from parent User.Then adds its own unique step — submit task!

6. Another child class **Manager** — also inherits from User.Gets login and view_tasks 
* work() is Manager's own METHOD. It calls login and view_tasks from parent User.Then adds its own unique step — assign task!

7. OBJECTS CREATED and CALLING METHODS

| Term | What it is | Example here |
|---|---|---|
| **Class** | Blueprint | `class User` |
| **Object** | Real thing created from class | `user1`, `user2` |
| **Attribute** | Data stored inside object | `self.name` → `"Aishwarya_M"` |
| **Method** | Action the object can do | `login()`, `work()` |
| **Inheritance** | Child gets parent's methods free | `TeamMember(User)` |



* Step1: Create a class
