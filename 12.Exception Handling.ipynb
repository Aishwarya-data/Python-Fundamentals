#### Exception handling is used to handle runtime errors and prevent program crash.
Why exceptions ?

1. To prevent program crash and handle errors
2. It gives user friendly output and makes debug easier

## 1.try and except



Syntax

try:

    risky code
    
except:

    error handling code

```python
try:
    num = int(input("Enter number: "))
    print(10 / num)

except:
    print("Error occurred")
```

#### Specific Exception Handling

```python
try:
    a = int(input("Enter number: "))
    b = int(input("Enter number: "))
    print(a/b)

except ZeroDivisionError:
    print("Cannot divide by zero")

except ValueError:
    print("Invalid input")
```

## 2.try except else

try:

    code
    
except:

    error code
    
else:

    success code

```python
try:
    num = int(input("Enter number: "))
    print(10 / num)

except ZeroDivisionError:
    print("Cannot divide by zero")

else:
    print("Operation successful")
```

## 3.try except finally

```python
try:
    num = int(input("Enter number: "))
    print(10 / num)

except ZeroDivisionError:
    print("Cannot divide by zero")

finally:  # finally block always execute even if error occurs
    print("Execution completed")
```

## 4.Raising Exceptions

```python
age = int(input("Enter age: "))

if age < 18:
    raise Exception("Age must be 18 or above")
else:
    print(age)
```

```python
try:
    age = int(input("Enter age: "))
    print("Age is:", age)

except ValueError:
    print("Please enter a valid number")
```

```python
try:
    a = int(input("Enter number: "))
    b = int(input("Enter number: "))
    print(a/b)

except ZeroDivisionError:
    print("Division by zero not allowed")

except ValueError:
    print("Invalid input")

finally:
    print("Program finished")
```

| Exception         | Meaning                  |
| ----------------- | ------------------------ |
| ZeroDivisionError | division by zero         |
| ValueError        | wrong value type         |
| TypeError         | wrong data type          |
| IndexError        | index out of range       |
| KeyError          | dictionary key not found |
| FileNotFoundError | file not found           |
