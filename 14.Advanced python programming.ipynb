### Factors of a number

user enter 20, output should be "The facters are 1,2,4,5,10,20

```python
usr_input = int(input("Enter number:"))

result = []
for i in range(1,usr_input+1):
    if usr_input%i==0:
        result.append(i)
print("The factors of {} is {} .".format(usr_input,result))
        
```
