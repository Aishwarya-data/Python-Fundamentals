# 1.Print

### print() is a built-in Python function used to display output on the screen.

### SYNTAX: print(objects,sep=' ',end='\n')
* print() → output display

* sep → separator between objects

* end → line ending control

* strings always " " inside

#### PRINT TEXT

```python
print("I am a data scientist and I am earning 1 lakh per month")
```

```python
print("Hello","world!")
print("Hello "," world",sep = "*")
```

#### PRINT MULTIPLE VALUES

```python
print("Data scientist","chennai","1 Lakh")
name = "Aishwarya"
age = 25
salary = 1000000
Location = "Chennai"
print(name,age,salary,Location)
```

#### PRINT WITH MESSAGE

```python
print("Hello",name)
```

#### PRINT NUMBERS , PRINT MULTIPLE LINES

```python
print(15+24)
print("Aishwarya")
```

```python
name = "Aishwarya M"
salary = 1000000

print("Employee Name:", name)
print("Salary:", salary)

# formatted print
print(f"My name is {name} and I am {age} years old.I am earning {salary} ")
```
